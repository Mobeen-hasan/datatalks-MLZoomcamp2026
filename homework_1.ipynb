Q6. Median value of horsepower

```python
'''
    Find the median value of the horsepower column in the dataset.
    Next, calculate the most frequent value of the same horsepower column.
    Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
    Now, calculate the median value of horsepower once again.

Has it changed?

    Yes, it increased
    Yes, it decreased
    No

'''
# Find the median value of the horsepower column in the dataset.
horsepower_median = df['horsepower'].median()
print("Initial median:", horsepower_median)

# Next, calculate the most frequent value of the same horsepower column.
horsepower_mode = df['horsepower'].mode()[0]
print("Most frequent value:", horsepower_mode)

# Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
df['horsepower'] = df['horsepower'].fillna(horsepower_mode)

# Now, calculate the median value of horsepower once again.
new_horsepower_median = df['horsepower'].median()
print("New median:", new_horsepower_median)

# Has it changed?
if new_horsepower_median > horsepower_median:
    print("Yes, it increased")
elif new_horsepower_median < horsepower_median:
    print("Yes, it decreased")
else:
    print("No")
```
